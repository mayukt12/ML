```python
import matplotlib.pyplot as plt  
import pandas as pd 
import numpy as np 
```

```python
df = pd.read_csv('placement (2).csv')
```

```python
df.head()
```

```python
plt.scatter(data=df , x='cgpa',y='package')
```

```python
x = df.iloc[:,0:1]
y=df.iloc[:,-1]
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr = LinearRegression()
```

```python
lr.fit(x_train,y_train)
```

```python
plt.scatter(data=df , x='cgpa',y='package')
plt.plot(x_train,lr.predict(x_train),color='red')
plt.xlabel('cgpa')
plt.ylabel('package')
```

```python
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
```

```python
y_pred=lr.predict(x_test)
```

```python
y_test.values
```

# MEAN ABSOLUTE ERROR

```python
print('MAE',mean_absolute_error(y_test,y_pred))
```

# mean squared error

```python
print('MsE',mean_squared_error(y_test,y_pred))
```

# root mean squared error

```python
print('RMSE',np.sqrt(mean_squared_error(y_test,y_pred)))
```

# r2 score

```python
print('R2',r2_score(y_test,y_pred))
```

```python
x_test.shape
```

# adjusted r2 score

```python
r2 = r2_score(y_test,y_pred)

1- ((1-r2)*(40-1)/(40-1-1))
```

```python
new_df =df.copy()
```

```python
new_df['random_feature']=np.random.random(200)

new_df=new_df[['cgpa','random_feature','package']]
```

```python
new_df.head()
```

```python
plt.scatter(data=new_df,x='random_feature',y='package')
```

```python
x = new_df.iloc[:,0:2]
y=new_df.iloc[:,-1]
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
from sklearn.linear_model import LinearRegression
lr = LinearRegression()

lr.fit(x_train,y_train)
```

```python
y_pred=lr.predict(x_test)
print('R2',r2_score(y_test,y_pred))
r2 = r2_score(y_test,y_pred)

print( 'adjusted r2',1- ((1-r2)*(40-1)/(40-1-1)))
```
