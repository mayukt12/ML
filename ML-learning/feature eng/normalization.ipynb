#  normalization is used when data has strict boundries like image with 255 to 0  pixel  or 0 to 1 

```python
import numpy as np 
import pandas as pd 
import seaborn as sns 
import matplotlib.pyplot as plt 
```

```python
df = pd.read_csv('wine_data.csv',header=None,usecols=[0,1,2])
df.columns = ['class label', 'alcohol', 'malic acid']
```

```python
df
```

```python
sns.kdeplot(df['malic acid'])
```

```python
sns.kdeplot(df['alcohol'])
```

```python
sns.scatterplot(x='alcohol',y='malic acid',hue='class label' , data=df , palette={1:'green',2:'blue',3:'red'})
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test = train_test_split(df.drop('class label',axis = 1),df['class label'],test_size=0.3,random_state=42)
```

```python
x_train.sample(5)
```

```python
x_train.shape , x_test.shape
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
scaler = MinMaxScaler()
```

```python
x_train_scaler=scaler.fit_transform(x_train)
x_test_scaler = scaler.transform(x_test)
```

```python
x_train_scaler = pd.DataFrame(x_train_scaler,columns=x_train.columns)
x_test_scaler = pd.DataFrame(x_test_scaler , columns = x_test.columns)
```

```python
np.round(x_train_scaler.describe(),1)
```

```python
np.round(x_test_scaler.describe(),1)
```

```python
fig, (ax1,ax2)= plt.subplots(ncols=2, figsize=((12,5)))
ax1.scatter(x_train['alcohol'],x_train['malic acid'],c= y_train)
ax1.set_title('before scaler')
ax2.scatter(x_train_scaler['alcohol'],x_train_scaler['malic acid'],c= y_train)
ax2.set_title('after scaler')
```

```python
fig, (ax1,ax2)= plt.subplots(ncols=2, figsize=((12,5)))
ax1.set_title("before scaling")
sns.kdeplot(x_train['alcohol'],ax=ax1)
sns.kdeplot(x_train['malic acid'],ax=ax1)

ax2.set_title("after scaling")
sns.kdeplot(x_train_scaler['alcohol'],ax=ax2)
sns.kdeplot(x_train_scaler['malic acid'],ax=ax2)

```

```python
fig, (ax1,ax2)= plt.subplots(ncols=2, figsize=((12,5)))
ax1.set_title("before scaling")
sns.kdeplot(x_train['alcohol'],ax=ax1)


ax2.set_title("after scaling")
sns.kdeplot(x_train_scaler['alcohol'],ax=ax2)

```

```python
fig, (ax1,ax2)= plt.subplots(ncols=2, figsize=((12,5)))
ax1.set_title("before scaling")
sns.kdeplot(x_train['malic acid'],ax=ax1)
ax2.set_title("after scaling")
sns.kdeplot(x_train_scaler['malic acid'],ax=ax2)
```
