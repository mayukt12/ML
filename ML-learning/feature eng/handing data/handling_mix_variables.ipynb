```python
import numpy as np 
import pandas as pd 
```

```python
df = pd.read_csv('titanic.csv')
```

```python
df.head()
```

```python
fig = df['number'].value_counts().plot.bar()
```

```python
# extract numeric and categorical part 

df['number_numeric'] = pd.to_numeric(df['number'],errors='coerce',downcast='integer')
df['number_categorical'] = np.where(df['number_numeric'].isnull(),df['number'],np.nan )
```

```python
df.head()
```

```python
df['cab_num'] = df['Cabin'].str.extract('(\d+)') # capture numeric part
df['cab_cat'] = df['Cabin'].str[0] # capture first letter
```

```python
df.head()
```

```python
df['cab_cat'].value_counts().plot(kind='bar')
```

```python
df['ticket_num'] = df['Ticket'].apply(lambda s: s.split()[-1])
df['ticket_num'] = pd.to_numeric(df['ticket_num'],errors='coerce',downcast='integer')

df['cat_ticket'] = df['Ticket'].apply(lambda s: s.split()[0])
df['cat_ticket'] = np.where(df['cat_ticket'].str.isdigit(), np.nan , df['cat_ticket'] )
```
