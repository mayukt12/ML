```python
import numpy as np 
import pandas as pd 
```

```python
time = pd.read_csv('messages.csv')
date = pd.read_csv('orders.csv')
```

```python
date.head()
```

```python
date.info()
```

```python
date['date'] = pd.to_datetime(date['date'])
```

```python
date.info(
    
)
```

```python
date['date_year'] = date['date'].dt.year
date.head()
```

```python
date['date_day'] = date['date'].dt.day
date.head()
```

```python
date['date_month'] = date['date'].dt.month
date.head()
```

```python
date['date_month_name'] = date['date'].dt.month_name()
date.head()
```

```python
date['date_day_name'] = date['date'].dt.day_name()
date.head()
```

```python
date['date_day_no'] = date['date'].dt.dayofweek
date.head()
```

```python
date['date_is_weekend'] = np.where(date['date_day_name'].isin(['Saturday','Sunday']),1,0)
date.head()
```

```python
# extract week no 
date['date_week_no'] = date['date'].dt.isocalendar().week
date.head()
```

```python
date = date.drop(columns=['product_id','city_id','orders'])
```

```python
date['date_quater'] = date['date'].dt.quarter
date.head()
```

```python
date['semester'] = np.where(date['date_quater'].isin([1,2]),1,2)
date.head()
```

```python
import datetime 
today = datetime.datetime.today()
```

```python
today- date['date']
```

```python
(today- date['date']).dt.days
```

```python
np.round((today- date['date']).dt.days / 30.4375,0)
```
