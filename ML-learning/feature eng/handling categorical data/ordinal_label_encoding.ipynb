# ordinal encoding is done when data is related like education 


```python
import pandas as pd 
import numpy as np 
```

```python
df = pd.read_csv('customer.csv')
```

```python
df.head()
```

#  here gender is an nominal encoding because it cant be compare 

```python
df = df.iloc[:,2:]
```

```python
df.head()
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test = train_test_split(df.iloc[:,0:2],df.iloc[:,-1] , test_size=0.2,random_state=42)
```

```python
x_train.head()
```

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
oe = OrdinalEncoder(categories=[['Poor','Average','Good'],['School','UG','PG']])
```

```python
oe
```

```python
x_train_oe = oe.fit_transform(x_train)
```

```python
x_train_oe = pd.DataFrame(x_train_oe,columns = x_train.columns)
```

```python
x_train_oe.head()
```

```python
x_test_oe = oe.transform(x_test )
```

```python
x_test_oe = pd.DataFrame(x_test_oe,columns=x_test.columns)
```

```python
x_test_oe
```

```python
x_test
```
