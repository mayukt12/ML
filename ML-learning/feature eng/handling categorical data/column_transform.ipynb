```python
import numpy as np 
import pandas as pd 
```

```python
df = pd.read_csv('covid_toy.csv')
```

```python
df.sample(5)
```

```python
df.isnull().sum()
```

```python
from sklearn.impute import SimpleImputer
```

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test ,y_train,y_test = train_test_split(df.drop(columns=['has_covid']),df['has_covid'],test_size=0.2,random_state=42)
```

```python
x_train.sample(5)
```

# without column transformation 

```python
si = SimpleImputer()
```

```python
# fill the null value by mean 
x_train_fever = si.fit_transform(x_train[['fever']])
x_test_fever = si.fit_transform(x_test[['fever']])

```

```python
x_train_fever.shape
```

```python
oe = OrdinalEncoder(categories=[['Mild','Strong']])
```

```python
x_train_cough = oe.fit_transform(x_train[['cough']])
x_test_cough = oe.fit_transform(x_test[['cough']])
```

```python
x_train_cough.shape
```

```python
ohe = OneHotEncoder(drop='first',sparse_output=False)
```

```python
x_train_ohe = ohe.fit_transform(x_train[['gender','city']])
x_test_ohe = ohe.fit_transform(x_test[['gender','city']])
```

```python
x_train_age = (x_train['age'].values).reshape(-1,1)
```

```python
x_test_age = x_test['age'].values.reshape(-1,1)
```

```python
x_train_all = np.concatenate((x_train_age,x_train_ohe,x_train_fever,x_train_cough) , axis = 1)
x_test_all = np.concatenate((x_test_age,x_test_ohe,x_test_fever,x_test_cough) , axis = 1)
```

# with column transform

```python
from sklearn.compose import ColumnTransformer
```

```python
# remainder='passthrough' means to not to change other column  
#remainder='drop' means to drop other columns 

tr = ColumnTransformer(transformers=[
    ('trf1',SimpleImputer(),['fever']),
    ('trf2',OrdinalEncoder(categories=[['Mild','Strong']]),['cough']), 
    ('trf3',OneHotEncoder(drop='first',sparse_output=False),['gender','city'])
],remainder='passthrough')
```

```python
tr.fit_transform(x_train).shape
```

```python
tr.transform(x_test)
```
