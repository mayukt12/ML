```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
tips = sns.load_dataset(tips)
```

# getting data how many peoples go to restaurant on whichtime 

```python
tps=tips.groupby(['time','sex'],observed = False)['size'].sum().reset_index()
```

```python
sns.barplot(data = tps ,x = 'time',y= 'size',hue='sex')
```

```python
# Create a grid of box plots for all 4 categorical columns
fig, axes = plt.subplots(2, 2, figsize=(12, 10))

# Sex vs Tip
sns.boxplot(ax=axes[0, 0], data=tips, x="sex", y="tip")
axes[0, 0].set_title("Sex vs Tip")

# Smoker vs Tip
sns.boxplot(ax=axes[0, 1], data=tips, x="smoker", y="tip")
axes[0, 1].set_title("Smoker vs Tip")

# Day vs Tip
sns.boxplot(ax=axes[1, 0], data=tips, x="day", y="tip")
axes[1, 0].set_title("Day vs Tip")

# Time vs Tip
sns.boxplot(ax=axes[1, 1], data=tips, x="time", y="tip")
axes[1, 1].set_title("Time vs Tip")

plt.tight_layout()
plt.show()
```
